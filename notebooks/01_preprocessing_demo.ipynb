This notebook is for the FaceForensics++ robustness project. Start with c23 originals and Deepfakes.

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
from src.transformations import jpeg_compress, center_crop, resize_roundtrip, gaussian_blur, contrast_filter, screenshot_like
```

## Dataset checks
After FaceForensics++ is downloaded, set `REAL_DIR` and `FAKE_DIR` to extracted frame folders.

```python
REAL_DIR = Path('data/frames/real')
FAKE_DIR = Path('data/frames/fake')
real_images = sorted(REAL_DIR.glob('*.jpg')) + sorted(REAL_DIR.glob('*.png'))
fake_images = sorted(FAKE_DIR.glob('*.jpg')) + sorted(FAKE_DIR.glob('*.png'))
print('real frames:', len(real_images))
print('fake frames:', len(fake_images))
```

## Transformation preview
Use a held-out frame here once the dataset is available.

```python
sample_path = real_images[0] if real_images else Path('evidence/generated_test_image.png')
im = Image.open(sample_path).convert('RGB')
variants = {
    'baseline': im,
    'jpeg_q35': jpeg_compress(im, 35),
    'crop_80': center_crop(im, 0.80),
    'resize_50': resize_roundtrip(im, 0.50),
    'blur_2': gaussian_blur(im, 2.0),
    'contrast_1.35': contrast_filter(im, 1.35),
    'screenshot_proxy': screenshot_like(im),
}
for name, image in variants.items():
    print(name, image.size)
```
